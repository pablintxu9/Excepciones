**Identificar los errores en los siguiente bloques de código y evaluarlos con excepciones especificas para evitar errores no controlados en nuestros programas. Añadem mensajes explicativos para el usuario.**
<br>_Nota: Se tienen que evaluar excepciones concretas, no hacer referencia a Exception sin más._

**1) Código a evaluar:**

```python
try:
    numero = 7 / 0
except ZeroDivisionError:
    numero = None  
    print("Error: División por cero. No se puede dividir entre 0. Elija un denominador distinto de 0.")
else:
    print("Resultado:", numero)
```

**2) Código a evaluar:**

```python
try:
    lista = [4, 7, 30, 23, 5]
    valor = lista[10]
except IndexError:
    print(f"Error: Índice fuera de rango. La lista tiene {len(lista)} elementos; el índice 10 no existe.")
    print(f"Solución: usar un índice entre 0 y {len(lista)-1}.")
    valor = None
else:
    print("Elemento en índice 10:", valor)
```

**3) Código a evaluar:**

```python
paises = { "españa":"español", "eeuu":"inglés", "italia":"italiano" } 
try:
    idioma = paises["alemania"]
except KeyError as e:
    idioma = None
    print(f"Error: clave no encontrada ({e}). Claves disponibles: {list(paises.keys())}.")
    print("Solución: use una clave válida del diccionario (por ejemplo 'españa', 'eeuu' o 'italia').")
else:
    print("Idioma:", idioma)
```

**4) Localiza el error en el siguiente bloque de código.  Crea una excepción para evitar que el programa se bloquee y además explica en un mensaje al usuario la causa y/o solución:**

```python
try:
    resultado = "2" + 10
except TypeError as e:
    resultado = None
    print(f"Error: no se pueden operar tipos distintos ({e}).")
    print("Causa: intentar concatenar una cadena con un entero.")
    print("Solución: convertir los operandos al mismo tipo, por ejemplo:")
    print(" - Como cadena: '2' + str(10) ->", "'2' + str(10)")
    print(" - Como número: int('2') + 10 ->", int("2") + 10)
else:
    print("Resultado:", resultado)
```
